# MERGING JOINING CONCATENATION

Merging 2 Dataframe

```python
employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})
```

```python
print(employees)
salaries
```

```python
pd.merge(employees , salaries , on='employee_id' , how="outer")
```

# GROUP BY AND AGGREGATION

GROUP BY

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}
df8 = pd.DataFrame(data)
```

```python
df8
```

```python
cat = df8.groupby('Category')
for i , v in cat:
    print(i)
    print(v)
```

```python
cat = df8.groupby('Category')['Sales'].sum()
cat
```

```python
cat = df8.groupby('Store')['Sales'].sum()
cat
```

```python
cat = df8.groupby(['Category' , 'Store'])['Sales'].sum()
cat
```

Aggregation

```python
df8['Sales'].mean()
```

```python
df8['Sales'].agg(['mean' , 'sum' , 'median' , 'min' , 'max'])
```

# PIVOT TABLES


```python
data = {
    'Date': pd.date_range('2023-01-01', periods=20),
    'Product': ['A', 'B', 'C', 'D'] * 5,
    'Region': ['East', 'West', 'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West',
               'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South'],
    'Sales': np.random.randint(100, 1000, 20),
    'Units': np.random.randint(10, 100, 20),
    'Rep': ['John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary',
            'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice']
}
df9 = pd.DataFrame(data)
df9['Month'] = df9['Date'].dt.month_name()
df9['Quarter'] = 'Q' + df9['Date'].dt.quarter.astype(str)
df9
```

```python
pd.pivot_table(df9 , values='Sales' , index='Region' , columns='Product' , aggfunc='median')
```

```python
pivot2 = pd.pivot_table(df9, values=['Sales', 'Units'], index='Region', columns='Product')
pivot2
```

CrossTab

```python
pd.crosstab(df9['Region'],df9['Product'])
```
