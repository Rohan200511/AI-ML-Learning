```python
names = ["A", "B", "C"]
marks = [80, 90, 75]

print(list(zip(names , marks)))
```

```python
names = ["A", "B", "C"]
marks = [80, 90, 75]

for name , mark in zip(names , marks):
    print(name , "->" , mark)
```

```python
nums = [10, 20, 30, 40]

for index , num in enumerate(nums):
    print(index , "->" , num)
```

```python
nums = [2, 4, 7, 8]

print(any(x % 2 == 1 for x in nums))
```

```python
words = ["HELLO", "WORLD", "Python"]

print(all(s.isupper for s in words))
```

```python
words = ["apple", "kiwi", "banana", "fig"]

temp = sorted(words , key = lambda s : len(s))

print(temp)
```

```python
nums = [-10, 3, -2, 8, -1]
temp = sorted(nums , key = abs)
print(temp)
```

```python
students = [
    ("Rohan", 91),
    ("Aman", 76),
    ("Rahul", 88),
    ("Karan", 95)
]

temp = sorted(students , key = lambda x : x[1] , reverse = True)
print(temp)
```

```python
names = ["A", "B", "C"]
marks = [45, 88, 67]

print(dict(zip(names , marks)))
```

```python
nums = [12, 25, 18, 7, 40]

for index , val in enumerate(nums):
    if(val > 20):
        print(index)
```

```python
words = ["apple", "bat", "umbrella", "sky", "orange"]
vowels = ['a' , 'e' , 'i' , 'o' , 'u']

print(any(x for x in words if x[0] in vowels))
```

```python
words = ["AI", "ML", "DSA", "SQL"]

print(all(x for x in words if len(x) >= 2))
```

```python
names = ["Aman", "Rohan", "Rahul"]
marks = [75, 92, 85]

zipped = list(zip(names , marks))

temp = sorted(zipped , key = lambda x : x[1] , reverse=True)
print(temp)
```
