```python
#Squares of 1–10
squares = [x * x for x in range(1 , 11)]
print(squares)
```

```python
#Even numbers
result = ["Even" if x % 2 == 0 else "ODD" for x in squares]
print(result)
print(result.count("Even"))
```

```python
'''Length of each word
Remove spaces from a string
Convert all names to lowercase'''
names = ["Rohan" , "Rahul" , " Shivam " , "Aakash" , "Vaibhav"]
```

```python
print([len(x) for x in names])
print([x.strip() for x in names])
print([len(x) for x in names])
print([x.lower() for x in names])
```
