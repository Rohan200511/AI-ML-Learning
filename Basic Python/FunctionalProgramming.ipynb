```python
from functools import reduce
```

```python
arr = [2, 4, 6, 8]
print(list(map(lambda x : x * 2 , arr)))
```

```python
nums = [5, 8, 11, 14, 17, 20]
print(list(filter(lambda x : x % 2 == 1 , nums)))
```

```python
arr1 = ["python", "java", "c++", "go"]
print(list(map(lambda x : x.upper() , arr1)))
```

```python
nums1 = [-10, 5, 0, 8, -3, 12]
print(list(filter(lambda x : x > 0 , nums1)))
```

```python
my_list = [2, 3, 4, 5]

print(reduce(lambda x , y : x * y , my_list))
```

```python
string = [" apple ", " banana ", " mango "]
print(list(map(lambda x : x.strip() , string)))
```

```python
my_str = ["cat", "elephant", "dog", "giraffe", "ox"]
print(list(filter(lambda x : len(x) > 3 ,my_str)))
```

```python
my = ["AI", "Machine", "Learning", "Python"]
print(reduce(lambda x , y : x if len(x) > len(y) else y , my))
```

```python
given = [1, 2, 3, 4, 5, 6]

squared = list(map(lambda x : x ** 2 , given))

print(list(filter(lambda x : x % 2 == 0 , squared)))
```

```python
num = [1, 2, 3, 4, 5]
doubled = list(map(lambda x : x * 2 , num))

print(reduce(lambda x , y : x + y , doubled))
```
