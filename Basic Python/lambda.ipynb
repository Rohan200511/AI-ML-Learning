```python
squaree = lambda x : x * x

print(squaree(2))
```

```python
maxi = lambda a , b : max(a , b)

print(maxi(2 , 3))
```

```python
lisst_ = [15 , 89 , 20 , 1 , 65 , 3]
temp = sorted(lisst_ , reverse = True)
print(temp)
```

```python
lisst__ = [15 , 89 , 20 , 1 , 65 , 3]
temp = sorted(lisst__ , key = lambda x : -x)
print(temp)
```

```python
fruits = ["apple", "banana", "cherry" , "abc"]
print(sorted(fruits , key = lambda x : len(x))) #Ascending
print(sorted(fruits , key = lambda x : -len(x))) #Descending
```

```python
students = [("Rohan", 21), ("Aman", 19), ("Karan", 20)]
temp = sorted(students , key = lambda x : x[1])
print(temp)

temp = sorted(students , key = lambda x : -x[1])
print(temp)
```

```python
students = [
    {"name": "Rohan", "marks": 85},
    {"name": "Aman", "marks": 92},
    {"name": "Karan", "marks": 78}
]

temp = sorted(students , key = lambda x : x["marks"])
print(temp)
```
