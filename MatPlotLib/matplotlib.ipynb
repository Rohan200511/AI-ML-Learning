```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0 , 5 , 11)
x
```

```python
y = x ** 2
y
```

```python
plt.plot(x , y)
plt.title("Hello")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.show()
```

# Subplots

```python
plt.subplot(2 , 2 , 1)
plt.plot([1 , 2 , 3] , [1 , 4 , 9])
plt.subplot(2 , 2 , 4)
plt.plot(x , y)
plt.subplot(2 , 2 , 3)
plt.plot(y , x)
plt.subplot(2 , 2 , 2)
plt.plot(x ** 2 , x ** 2)
```

```python
plt.subplot(6 , 6 , 1)
plt.subplot(6 , 6 , 20)
plt.subplot(6 , 6 , 36)
```

# Object oriented Way

```python
fig = plt.figure(dpi=120)
axis1 = fig.add_axes([0.1 , 0.1 , 1, 1])
axis1.plot(x , y)

axis2 = fig.add_axes([0.2 , 0.4 , 0.3 , 0.3])
axis2.plot(x , y)
```

# TYPES OF PLOTS

```python
plt.scatter(x , y)
plt.show()
```

```python
plt.boxplot(x)
plt.show()
```

```python
fig = plt.figure()
ax = fig.add_axes([0.1 , 0.1 , 0.8 , 0.8])
ax.plot(x , y)
plt.savefig("Basic Plot.png")
```
