```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('IPL.csv')
df.head(5)
```

Which team won the most matches?

```python
match_wins = df['match_winner'].value_counts()
sns.barplot(y = match_wins.index , x = match_wins.values , palette='viridis')
plt.title("Most Match win by team")
```

Toss Decision Trends

```python
sns.countplot(x = df['toss_decision'])
plt.title("Toss Decision Trends")
```

Toss Winner vs Match Winner

```python
count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
precentage = (count / 74 )* 100
precentage
```

How do teams win? (Runs vs Wickets)

```python
sns.countplot(x = df['won_by'])
plt.title("Won By: ")
```

1 Most "Player of the Match" Awards

```python
df.head(2)
```

```python
countt = df['player_of_the_match'].value_counts().head(10)
countt
```

```python
sns.barplot(x = countt.index , y = countt.values , palette='mako')
plt.title("Top 10 Players with Man of the Match")
plt.xticks(rotation=90)
plt.show()
```

2 Top Scorers

```python
df.head(2)
```

```python
high = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)
high
```

```python
high.plot(kind = 'bar')
```

10 Best Bowling Figures

```python
df.head(2)
```

```python
df['highest_wickets'] = df['best_bowling_figure'].apply(lambda x : x.split('--')[0])
df['highest_wickets'] = df['highest_wickets'].astype(int)
top_bowlers = df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False).head(10)
top_bowlers
```

```python
top_bowlers.plot(kind = 'bar')
```

Most Matches Played by Venue

```python
venue_count = df['venue'].value_counts()
venue_count
```

```python
sns.barplot(x = venue_count.index , y = venue_count.values)
plt.xticks(rotation = 90)
```
