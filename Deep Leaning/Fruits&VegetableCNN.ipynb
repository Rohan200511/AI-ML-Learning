```python
import os
os.listdir("train")
```

```python
train_ds = tf.keras.utils.image_dataset_from_directory(
    "/content/train",
    image_size=(224, 224),
    batch_size=32,
    label_mode='categorical'
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    "/content/validation",
    image_size=(224, 224),
    batch_size=32,
    label_mode='categorical'
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    "/content/test",
    image_size=(224, 224),
    batch_size=32,
    shuffle=False,
    label_mode='categorical'
)
```

```python
class_names = train_ds.class_names
len(class_names)
```

```python
train_ds
```

```python
from tensorflow.keras.layers import Flatten , Rescaling , Dropout
```

```python
model = Sequential([
    tf.keras.layers.Rescaling(1./255),
    Conv2D(32 , kernel_size=(3 , 3) , activation='relu' , input_shape=(224 , 224 , 3)),
    MaxPooling2D(pool_size = (2 , 2)),
    Conv2D(64 , kernel_size=(3 , 3) , activation='relu'),
    MaxPooling2D(pool_size=(2 , 2)),
    Conv2D(128 , kernel_size=(3 , 3) , activation='relu'),
    MaxPooling2D(pool_size=(2 , 2)),
    Flatten(),
    Dropout(0.5),
    Dense(128 , activation = 'relu'),
    Dense(36 , activation='softmax')
])
```

```python
model.compile(optimizer='adam' , loss = 'categorical_crossentropy' , metrics=['accuracy'])
```

```python
history = model.fit(train_ds , validation_data = val_ds , epochs = 10 , batch_size = 32 , verbose = 1)
```

```python
accuracy = model.evaluate(test_ds , verbose = 1)[1]
accuracy
```

```python
from tensorflow.keras.preprocessing import image
```

```python
img_1 = 'apple.jpg'
img_2 = 'd.jpg'
target_size = (224 , 224)

img1 = image.load_img(img_1 , target_size = target_size)
img2 = image.load_img(img_2 , target_size = target_size)
```

```python
img1_array = image.img_to_array(img1)
img2_array = image.img_to_array(img2)
```

```python
img1_array = np.expand_dims(img1_array , axis = 0)
img2_array = np.expand_dims(img2_array , axis = 0)
img1_array = img1_array / 255.0
img2_array = img2_array / 255.0
```

```python
pred1 = model.predict(img1_array)
print(pred1)

pred2 = model.predict(img2_array)
print(pred2)
```

```python

predicted_class_index_1 = np.argmax(pred1[0])
predicted_class_name_1 = class_names[predicted_class_index_1]
confidence_1 = np.max(pred1[0]) * 100

print(f"For '{img_1}':")
print(f"Predicted Class: {predicted_class_name_1}")
print(f"Confidence: {confidence_1:.2f}%")


predicted_class_index_2 = np.argmax(pred2[0])
predicted_class_name_2 = class_names[predicted_class_index_2]
confidence_2 = np.max(pred2[0]) * 100

print(f"\nFor '{img_2}':")
print(f"Predicted Class: {predicted_class_name_2}")
print(f"Confidence: {confidence_2:.2f}%")
```
